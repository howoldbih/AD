# 8)

* **MEDV** and **LSTAT**: Сильная отрицательная зависимость. Чем меньше процент населения с низким социальным статусом, тем выше цена недвижимости.

* **MEDV** and **RM**: Сильная положительная зависимость. Чем больше кол-во комнат в недвижимости, тем больше ее цена.

* **MEDV** and **PTRATIO**: Малая отрицательная зависимость. Чем меньше соотношение кол-ва учеников и учителей, тем больше недвижимости.

* **MEDV** and **INDUS**: Малая отрицательная зависимость. Чем меньше доля площадей, не связанных с розничной торговлей, тем выше цена недвижимости.

* **MEDV** and **TAX**: Малая отрицательная зависимость. Чем меньше налоги, тем выше цена недвижимости.

# 15)

```python
df_wo_outliers = df[df['MEDV'] < 37]

X_f = df_wo_outliers[top5_features_list]
y_f = df_wo_outliers['MEDV']

X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(X_f, y_f, test_size=0.2, random_state=42)

model_f = LinearRegression()
model_f.fit(X_train_f, y_train_f)

y_train_prediction_f = model_f.predict(X_train_f)
y_test_prediction_f = model_f.predict(X_test_f)

r2_train_score_f = r2_score(y_train_f, y_train_prediction_f)
rmse_train_score_f = np.sqrt(mean_squared_error(y_train_f, y_train_prediction_f))
r2_test_score_f = r2_score(y_test_f, y_test_prediction_f)
rmse_test_score_f = np.sqrt(mean_squared_error(y_test_f, y_test_prediction_f))

print("Значения ошибок на данные без выбросов в признаке MEDV")
print(f"Значение R^2 ошибки на тренировочную выборку: {r2_train_score_f}")
print(f"Значение RMSE ошибки на тренировочную выборку: {rmse_train_score_f}")
print(f"Значение R^2 ошибки на тестовую выборку: {r2_test_score_f}")
print(f"Значение RMSE ошибки на тестовую выборку: {rmse_test_score_f}")
```

Сравнения:
* R2 ошибка увеличилась c 0.6196 до 0.6888
* RMSE ошибка уменьшилась с 5.2812 до 3.1591

Вывод:
* Линейная регрессия лучше работает с данными без выбросов.
* Порой лучше избавиться от выбросов при обучении модели.

# 17)

```python
poly = PolynomialFeatures(degree=3)

X_poly_train = poly.fit_transform(X_train_f)
X_poly_test = poly.fit_transform(X_test_f)

model_poly = LinearRegression()
model_poly.fit(X_poly_train, y_train_f)

y_train_prediction_f_p = model_poly.predict(X_poly_train)
y_test_prediction_f_p = model_poly.predict(X_poly_test)

r2_train_score_f_p = r2_score(y_train_f, y_train_prediction_f_p)
rmse_train_score_f_p = np.sqrt(mean_squared_error(y_train_f, y_train_prediction_f_p))
r2_test_score_f_p = r2_score(y_test_f, y_test_prediction_f_p)
rmse_test_score_f_p = np.sqrt(mean_squared_error(y_test_f, y_test_prediction_f_p))

print("Значения ошибок на данные без выбросов в признаке MEDV и с использованием полиномиальной регрессии 3-ей степени:")
print(f"Значение R^2 ошибки на тренировочную выборку: {r2_train_score_f_p}")
print(f"Значение RMSE ошибки на тренировочную выборку: {rmse_train_score_f_p}")
print(f"Значение R^2 ошибки на тестовую выборку: {r2_test_score_f_p}")
print(f"Значение RMSE ошибки на тестовую выборку: {rmse_test_score_f_p}")
```

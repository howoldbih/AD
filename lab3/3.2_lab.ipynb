# Импорт библиотек

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from scipy.linalg import solve
```

# Выгрузка датасета

```python
data = pd.read_csv('/content/data2.csv', sep=';', names=['discount', 'income'])
data.head()
```

# 1)

```python
x = data['discount']
y = data['income']
n = len(x)
ids_2polinom = [0, n//2 ,n-1]
x_dots_2polinom = x[ids_2polinom]
y_dots_2polinom = y[ids_2polinom]

for i in ids_2polinom:
  print(f"Выбранные точки: x={x[i]}, y={y[i]}")
```

# 2)

```python
x_expression_2polinom = np.vstack([x_dots_2polinom**2, x_dots_2polinom, np.ones(3)]).T
y_expression = y_dots_2polinom

coefs = solve(x_expression_2polinom, y_expression)
a2, a1, a0 = coefs

print(f"Коэффициенты СЛУ: a2={a2:.2f}, a1={a1:.2f}, a0={a0:.2f}")
```

# 3)

```python
y_pred_2polinom = a0 + a1*x + a2*(x**2)

print("Предугаданное значение:")
print(y_pred_2polinom)
```

# 4)

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='Исходные данные'))
fig.add_trace(go.Scatter(x=x, y=y_pred_2polinom, mode='lines+markers', name='Полином 2-ой степени'))
fig.update_layout(xaxis_title='Скидка',
                  yaxis_title='Прибыль')
fig.show()
```

# 5)

```python
error_2polinom = np.sum((y - y_pred_2polinom)**2)

print("Квадратичное отклонение для полинома второй степени:", error_2polinom)
```

# 6.1)

```python
ids_3polinom = [0, 2, n//2 ,n-1]
x_dots_3polinom = x[ids_3polinom]
y_dots_3polinom = y[ids_3polinom]

for i in ids_3polinom:
  print(f"Выбранные точки: x={x[i]}, y={y[i]}")
```

# 6.2)

```python
x_expression_3polinom = np.vstack([x_dots_3polinom**3, x_dots_3polinom**2, x_dots_3polinom, np.ones(4)]).T
y_expression = y_dots_3polinom

coefs = solve(x_expression_3polinom, y_expression)
a3, a2, a1, a0 = coefs

print(f"Коэффициенты СЛУ: a3={a3:.2f}, a2={a2:.2f}, a1={a1:.2f}, a0={a0:.2f}")
```

# 6.3)

```python
y_pred_3polinom = a0 + a1*x + a2*(x**2) + a3*(x**3)

print("Предугаданное значение:")
print(y_pred_3polinom)
```

# 6.4)

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='Исходные данные'))
fig.add_trace(go.Scatter(x=x, y=y_pred_3polinom, mode='lines+markers', name='Полином 3-ей степени'))
fig.update_layout(xaxis_title='Скидка',
                  yaxis_title='Прибыль')
fig.show()
```

# 6.5)

```python
error_3polinom = np.sum((y - y_pred_3polinom)**2)

print("Квадратичное отклонение для полинома третьей степени:", error_3polinom)
```

# 8)

```python
def prediction(discount, a0, a1, a2, a3):
  return a0 + a1*discount + a2*(discount**2) + a3*(discount**3)
```

```python
print(f"Модель полинома 3-ей степени лучше, так как ошибка по сравнению с полиномом 2-ой степени меньше: {error_3polinom} < {error_2polinom}")

print("\nПрогноз прибыли для скидки 6%")
print(prediction(0.06, a0, a1, a2, a3))

print("\nПрогноз прибыли для скидки 8%")
print(prediction(0.08, a0, a1, a2, a3))
```

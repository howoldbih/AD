# Импорт библиотек

```python
import pandas as pd
import numpy as np
```

# 1)

```python
generation = pd.read_csv('/content/global-electricity-generation.csv')
consumption = pd.read_csv('/content/global-electricity-consumption.csv')

generation = generation.set_index('Country')
consumption = consumption.set_index('Country')

for col in generation.columns:
    generation[col] = pd.to_numeric(generation[col], errors='coerce')
for col in consumption.columns:
    consumption[col] = pd.to_numeric(consumption[col], errors='coerce')

print('Глобальное производство электроэнергии:')
display(generation.head())
print('Глобальное потребление электроэнергии:')
display(consumption.head())
```

# 2)

```python
generation_five_years = generation.iloc[:,-5:].mean()
consumption_five_years = consumption.iloc[:,-5:].mean()

print('Среднее глобальное производство электроэнергии за последние 5 лет:')
display(generation_five_years)
print('Среднее глобальное потребление электроэнергии за последние 5 лет:')
display(consumption_five_years)
```

# 3.1)

```python
total_consumption_per_year = consumption.sum(axis=0)

print("Суммарное потребление электроэнергии за каждый год")
display(total_consumption_per_year)
```

# 3.2)

```python
max_generation = generation.max().max()

print("Максимальное количество энергии произведенное за год:", max_generation)
```

# 3.3)

```python
mean_last_five_years_generation = generation.iloc[:,-5:].mean(axis=1)
generation_more_than_500 = mean_last_five_years_generation[mean_last_five_years_generation > 500]

print("Страны произодящие в среднем более 500 млрд. кВт*ч за последние 5 лет:")
display(generation_more_than_500)
```

# 3.4)

```python
mean_last_five_years_consumption = consumption.iloc[:,-5:].mean(axis=1)
quantile_90 = mean_last_five_years_consumption.quantile(0.9)
max_10_percent_consumption_countries = mean_last_five_years_consumption[mean_last_five_years_consumption > quantile_90]

print("10% стран, которые потребляют больше всего электроэнергии ежегодно в среднем за 5 лет:")
display(max_10_percent_consumption_countries)
```

# 3.5)

```python
counry_generation_with_10x_growth = generation[generation.loc[:, '2021'] / generation.loc[:, '1992'] > 10].index.tolist()

print("Список стран, которые увеличили производство электроэнергии в 2021 году по сравнению с 1992 годом более, чем в 10 раз:")
display([country.lstrip() for country in counry_generation_with_10x_growth])
```

# 3.6)

```python
mask_countries_consumption_more_100_less_than_generation = (consumption.sum(axis=1) > 100) & (generation.sum(axis=1) < consumption.sum(axis=1))
print("Список стран, которые в сумме за все годы потратили больше 100 млрд. кВт*ч электроэнергии и при этом произвели меньше, чем потратили:")
display([country.lstrip() for country in generation[mask_countries_consumption_more_100_less_than_generation].index.tolist()])
```

# 3.7)

```python
country_spent_most_in_2020 = consumption['2020'].idxmax().lstrip()
country_num_spent_most_in_2020 = consumption['2020'].max()
print(f"Страна, потратившая наибольшее количество электроэнергии в 2020 году: {country_spent_most_in_2020}, {country_num_spent_most_in_2020}")
```

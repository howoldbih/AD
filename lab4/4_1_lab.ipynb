# Иморт библиотек

```python
import numpy as np
import pandas as pd
```

# 1)

```python
M = 1
s = 1
n = 1000

np.random.seed(42)
series = pd.Series(np.random.normal(loc=M, scale=s, size=n))
print("Сгенерированный массив")
display(series)
```

# 2)

```python
numbers_in_range_1 = series[(series > M-s) & (series < M+s)].count()/n

print(f"Доля значений, находящихся  пределах [{M-s}; {M+s}] равна: {numbers_in_range_1} ")
```

# 3)

```python
numbers_in_range_2 = series[(series > M-3*s) & (series < M+3*s)].count()/n

print(f"Доля значений, находящихся  пределах [{M-3*s}; {M+3*s}] равна: {numbers_in_range_2} ")
print("Доля значений, получившаяся эмпирически, совпадает с теоретической")
```

# 4)

```python
sqrt_series = np.sqrt(series)

print("При попытке получить кадратный корень из значений, всплывает предупреждение из-за того, что в series присутствуют отрицательные значения, при попытке получить квадратный корень из таких чисел, numpy превращает их в nan значения")
print("Всего полученных nan значений, после попытки получить квадратный корень значений:", sqrt_series.isna().sum())
print(sqrt_series)
```

# 5)

```python
print("Среднее значение без nan:", sqrt_series.mean(skipna=True))
```

# 6)

```python
df = pd.DataFrame({"number": series, "root": sqrt_series})

print("Первые 6 строк созданного датафрейма:")
df.iloc[:6, :]
```

# 7)

```python
print("Значения в диапозоне [1.8; 1.9]\n")
df.query("root > 1.8 & root < 1.9")
```

```python
print("Статистическая информация: Age, Height, Weight")
display(df[['Age', 'Height', 'Weight']].describe())
```

# 4.1)

```python
id_1992 = df[df['Year'] == 1992]['Age'].idxmin()

age = df.iloc[id_1992]['Age']
name = df.iloc[id_1992]['Name']
discipline = df.iloc[id_1992]['Event']

print(f"Самому молодому участнику соревнования в 1992 году было {age} лет")
print(f"Имя самого молодого участника в 1992: {name}")
print(f"Дисциплина в которой участвовал самый молодой участник в 1992: {discipline}")
```

# 4.3)

```python
query_43 = df[(df['Year'] == 2000) & (df['Sex'] == "F") & (df['Sport'] == "Tennis")]["Height"].mean()

print("Средний рост теннисисток участвовавших в 2000:", query_43)
```

# 4.4)

```python
query_44 = df[(df["Year"] == 2008) & (df["Team"] == "China") & (df['Sport'] == "Table Tennis") & (df['Medal'] == 'Gold')]['Name'].count()

print("Кол-во золотых медалей заработанных Китаем в 1998 в настольном теннисе:", query_44)
```

# 4.5)

```python
sports_2004 = df[(df['Year'] == 2004) & (df['Season'] == 'Summer')]['Sport'].nunique()
sports_1988 = df[(df['Year'] == 1988) & (df['Season'] == 'Summer')]['Sport'].nunique()
print(f"На летних ОИ 2004 года было {sports_2004} видов спорта, а в 1988 - {sports_1988}.")
print(f"Изменение: {sports_2004 - sports_1988} видов спорта.")
```

# 4.6)

```python
query_46 = df[(df['Year'] == 2014) & (df['Sport'] == 'Curling') & (df['Sex'] == 'M')]['Age']

fig = go.Figure()
fig.add_trace(go.Histogram(x=query_46, nbinsx = 10))
fig.show()
```

# 4.7)

```python
query_46 = df[(df['Year'] == 2006) & (df['Season'] == 'Winter')]

country_stats = df.groupby('NOC').agg(MedalCount=('Medal', 'count'), AgeAverage=('Age', 'mean'))

print('Кол-во медалей и средний возраст по странам за 2006 год:')
display(country_stats[country_stats['MedalCount'] != 0])
```

# 4.8)

```python
pivot_table_2006 = pd.pivot_table(
    df,
    values='ID',
    index='NOC',
    columns='Medal',
    aggfunc='count',
    fill_value=0
)

print('Таблица по медалям за 2006 год:')
display(pivot_table_2006)
```

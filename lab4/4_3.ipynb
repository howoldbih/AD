# Иморт библиотек

```python
import pandas as pd
import numpy as np
```

# 1)

```python
df = pd.read_csv('/content/telecom_churn.csv')

display(df.describe())
display(df.info())
df
```

# 2)

```python
print('Сколько клиентов пользуется сервисом и не пользуется:')
print(df['Churn'].value_counts())
print('Сколько клиентов пользуется сервисом и не пользуется в долях:')
print(df['Churn'].value_counts(normalize=True))
```

# 3)

```python
df['Mean call duration'] = (df['Total day minutes'] / df['Total day calls'] + df['Total eve minutes'] / df['Total eve calls'] + df['Total night minutes'] / df['Total night calls']) / 3

print("Датафрейм с колонкой 'Mean call duration':")
display(df)
```

# 4)

```python
print("Среднее время звонка по значению поля Churn:")
print(df.groupby(['Churn'])['Mean call duration'].mean())
print("\nСущественная разница отсутстует")
```

# 5)

```python
avg_calls_service = df.groupby('Churn')['Customer service calls'].mean()
print("Среднее количество звонков в службу поддержки:")
print(avg_calls_service)
print(f"\nСущественная разница присутствует")
```

# 6)

```python
crosstab_service_calls = pd.crosstab(df['Customer service calls'], df['Churn'], normalize='index')

print("Таблица сопряженности Customer service calls и Churn:")
display(crosstab_service_calls)
print("\nОт четырех звонках процент оттока станоится значительным 45%")
```

# 7)

```python
crosstab_international_plan = pd.crosstab(df['International plan'], df['Churn'], normalize='index')

print("Таблица сопряженности International plan и Churn:")
display(crosstab_international_plan)
print("\nПроцент оттока пользователей при использования международного роуминга больше чем при не использовании")
```

# 8)

```python
df['Predicted churn'] = (df['International plan'] == 'Yes') | (df['Customer service calls'] >= 4)

false_positive = ((df['Churn'] == False) & (df['Predicted churn'] == True)).sum()
false_positive_rate = false_positive / (df['Churn'] == False).sum()

false_negative = ((df['Churn'] == True) & (df['Predicted churn'] == False)).sum()
false_negative_rate = false_negative / (df['Churn'] == True).sum()

print("Всего ложноположительных предсказаний:", false_positive)
print("Доля ложноположительных предсказаний:", false_positive_rate)
print("\nВсего ложноположительных предсказаний:", false_negative)
print("Доля ложнонегативных предсказаний:", false_negative_rate)
```
